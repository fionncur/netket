# Phase-separated model with two RBMs

Kernel environment must have fork installed,

`pip install -e .` from netket directory.

We are importing the PhaseSeparated model (from `netket/models/phase_separated`) to facilitate the use of arbitrary model inputs for the amplitude and phase parts of the wavefunction, primarily to use an RBM for the amplitude to allow Gibbs sampling, and compare arbitrary phase representations.

The existing `RBMModPhase` model from the library can be reproduced by passing two `RBM` models at construction. However, the implementation of `RBMModPhase` is equivalent only if the RBM objects have `use_visible_bias = False`, i.e. they are RBMs with $a_i = 0$ for all $i$. 

It is unclear why this decision was made, in fact it is inconsistent with the docstring for `RBMModPhase` which includes the visible bias layer in its `math` expression but does include it in its implementation. In any case, the PhaseSeparated object allows a visible bias layer. 

```python
import jax
import jax.numpy as jnp
import netket as nk

from netket.models import PhaseSeparated
```

Disable visible biases in both RBMs to match the architecture of `RBMModPhase`. Their parameters are initialized independently.


```python
amplitude = nk.models.RBM(alpha=1, use_visible_bias=False)
phase = nk.models.RBM(alpha=1, use_visible_bias=False)

PhaseSeparated = PhaseSeparated(amplitude=amplitude, phase=phase)
PhaseSeparated
```

```python
RBMModPhase = nk.models.RBMModPhase(alpha = 1)
RBMModPhase
```
